```python
import time
from os import walk
import os
from gui_code import VideoAnalysisModel as vam
import matplotlib.pyplot as plt

f = []
dir = os.getcwd()
clips = os.path.join(dir, "clips")

start_time = time.time()

filenames = next(walk(clips), (None, None, []))[2]
file_times = []  

print(filenames)
for file in filenames:
    print(f"Analyzing {file}...")
    file_start = time.time()  
    info1, info2, info3, info4, info5 = vam.analyze_video_model(
        os.path.join('clips', file), "analysed" + file
    )
    file_end = time.time()
    elapsed_time = file_end - file_start
    file_times.append((file, elapsed_time))  
    print(f"Time taken to analyze {file}: {elapsed_time:.2f} seconds")

end_time = time.time()
total_time = end_time - start_time
print(f"Total execution time: {total_time:.2f} seconds")

file_names = [ft[0] for ft in file_times]
times = [ft[1] for ft in file_times]

plt.figure(figsize=(12, 6))

plt.bar(file_names, times, color='blue', alpha=0.7)
plt.xlabel('Files')
plt.ylabel('Time Taken (seconds)')
plt.title('Processing Time Per File')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()


```
