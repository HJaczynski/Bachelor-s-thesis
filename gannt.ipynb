```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

data = {
    "Task": [
        "Project Topic Definition", "Project Details Planning", "Project Topic Description", 
        "Problem Analysis and Research", "Project Solution Description", 
        "Data Collection and Preparation / Setting up API", "Fine Tuning Existing Model", 
        "GUI Implementation", "Dashboard Implementation", "Model Implementation", 
        "Unit Testing", "Integration, System and Acceptance Testing", 
        "Project Evaluation and Final Documentation"
    ],
    "Start Date": [
        "2024-09-30", "2024-10-03", "2024-10-11", "2024-10-12", "2024-10-29", 
        "2024-11-08", "2024-11-11", "2024-11-14", "2024-11-18", "2024-11-21", 
        "2024-12-06", "2024-12-10", "2024-12-15"
    ],
    "End Date": [
        "2024-10-02", "2024-10-10", "2024-10-17", "2024-10-28", "2024-11-07", 
        "2024-11-10", "2024-11-13", "2024-11-30", "2024-11-25", "2024-12-05", 
        "2024-12-09", "2024-12-14", "2024-12-20"
    ]
}

df = pd.DataFrame(data)
df['Start Date'] = pd.to_datetime(df['Start Date'])
df['End Date'] = pd.to_datetime(df['End Date'])

fig, ax = plt.subplots(figsize=(10, 6))

for i, task in df.iterrows():
    ax.barh(task['Task'], (task['End Date'] - task['Start Date']).days, left=task['Start Date'])

ax.xaxis.set_major_locator(mdates.WeekdayLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %d'))

ax.set_xlabel('Date')
ax.set_ylabel('Tasks')
ax.set_title('Project Schedule Gantt Chart')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
